#Factorial Using Iterative Method

```python
import time

def factorial_iterative(n):
    fact = 1
    for i in range(1, n + 1):
        fact *= i
    return fact

# Get user input
n = int(input("Enter a number: "))

start_time = time.time()
result = factorial_iterative(n)
end_time = time.time()

execution_time = end_time - start_time

print(f"Factorial of {n} is: {result}")
print(f"Execution Time (Iterative): {execution_time:.6f} seconds")
print("Time Complexity (Iterative): O(N)")
print("Space Complexity (Iterative): O(1)")
```

#Factorial Using Recursive Method

```python
import time

def factorial_recursive(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial_recursive(n - 1)

# Get user input
n = int(input("Enter a number: "))

start_time = time.time()
result = factorial_recursive(n)
end_time = time.time()

execution_time = end_time - start_time

print(f"Factorial of {n} is: {result}")
print(f"Execution Time (Recursive): {execution_time:.6f} seconds")
print("Time Complexity (Recursive): O(N)")
print("Space Complexity (Recursive): O(N)")
```
