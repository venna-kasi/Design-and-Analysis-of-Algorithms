```python
def matrix_chain_order(p):
    n = len(p) - 1

    dp = [[0 for _ in range(n)] for _ in range(n)]

    for length in range(2, n + 1):
        for i in range(n - length + 1):
            j = i + length - 1

            dp[i][j] = float('inf')

            for k in range(i, j):

                cost = (
                    dp[i][k]
                    + dp[k + 1][j]
                    + p[i] * p[k + 1] * p[j + 1]
                )

                dp[i][j] = min(dp[i][j], cost)

    return dp


p = [1, 3, 2, 5, 4]

dp = matrix_chain_order(p)

print("DP Table:")

for row in dp:
    print(row)

print("Minimum number of multiplications =", dp[0][3])
```
