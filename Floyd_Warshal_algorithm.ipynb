#Floyd-Warshal algorithm.

```python
INF = float('inf')

graph = [
    [0, 5, INF, 10],
    [INF, 0, 3, INF],
    [4, INF, 0, 1],
    [7, INF, INF, 0]
]

n = len(graph)

for k in range(n):
    for i in range(n):
        for j in range(n):
            graph[i][j] = min(
                graph[i][j],
                graph[i][k] + graph[k][j]
            )

print("Shortest distance matrix:")

for row in graph:
    print(row)
```
