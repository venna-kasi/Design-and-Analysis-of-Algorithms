# Linear Search


```python
import time

def linear_search(arr, key):
    for i in range(len(arr)):
        if arr[i] == key:
            return i
    return -1

# Get user input
user_input = input("Enter numbers separated by spaces for Linear Search: ")

try:
    my_list = list(map(int, user_input.split()))
except ValueError:
    print("Invalid input. Please enter numbers only.")
    my_list = []

if my_list:
    key_input = input("Enter the element to search: ")
    try:
        # Attempt to get a single integer key. If multiple numbers are entered, take the first.
        key = int(key_input.split()[0])
    except (ValueError, IndexError):
        print("Invalid input for search key. Please enter a single number.")
        key = None # Set key to None to indicate invalid input

    if key is not None:
        print(f"Original list: {my_list}")

        # Measure execution time
        start_time = time.time()
        result = linear_search(my_list, key)
        end_time = time.time()

        execution_time = end_time - start_time

        if result != -1:
            print(f"Element {key} found at index {result}")
        else:
            print(f"Element {key} not found")

        print(f"Execution Time (Linear Search): {execution_time:.6f} seconds")
        print("Time Complexity (Linear Search):")
        print("  Best Case    : O(1)")
        print("  Average Case : O(N)")
        print("  Worst Case   : O(N)")
        print("Space Complexity (Linear Search): O(1)")
else:
    print("No valid numbers entered.")
```

# Binary Search


```python
import time

def binary_search(arr, key):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == key:
            return mid
        elif arr[mid] < key:
            low = mid + 1
        else:
            high = mid - 1

    return -1

# Get user input
user_input = input("Enter numbers separated by spaces for Binary Search: ")

try:
    my_list = list(map(int, user_input.split()))
except ValueError:
    print("Invalid input. Please enter numbers only.")
    my_list = []

if my_list:
    # Binary Search requires a sorted list
    my_list.sort()

    print(f"Sorted list: {my_list}")

    key = int(input("Enter the element to search: "))

    # Measure execution time
    start_time = time.time()
    result = binary_search(my_list, key)
    end_time = time.time()

    execution_time = end_time - start_time

    if result != -1:
        print(f"Element {key} found at index {result}")
    else:
        print(f"Element {key} not found")

    print(f"Execution Time (Binary Search): {execution_time:.6f} seconds")
    print("Time Complexity (Binary Search):")
    print("  Best Case    : O(1)")
    print("  Average Case : O(log N)")
    print("  Worst Case   : O(log N)")
    print("Space Complexity (Binary Search): O(1)")
else:
    print("No valid numbers entered.")
```
