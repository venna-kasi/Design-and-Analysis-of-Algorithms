```python


n = 4
capacity = 8

weight = [2, 3, 4, 5]
profit = [1, 2, 5, 6]


dp = [[0 for w in range(capacity + 1)] for i in range(n + 1)]

for i in range(1, n + 1):
    for w in range(1, capacity + 1):

        if weight[i - 1] <= w:
            dp[i][w] = max(
                dp[i - 1][w],
                profit[i - 1] + dp[i - 1][w - weight[i - 1]]
            )
        else:
            dp[i][w] = dp[i - 1][w]

print("DP Table:")

for row in dp:
    print(row)

print("Maximum Profit =", dp[n][capacity])
```
