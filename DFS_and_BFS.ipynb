# BSF


```python
from collections import deque

def bfs(graph, start):
    visited = set()
    queue = deque([start])

    visited.add(start)

    while queue:
        vertex = queue.popleft()
        print(vertex, end=" ")

        for neighbour in graph[vertex]:
            if neighbour not in visited:
                visited.add(neighbour)
                queue.append(neighbour)


# Graph representation
graph = {
    0: [1, 2],
    1: [0, 3, 4],
    2: [0, 5],
    3: [1],
    4: [1, 5],
    5: [2, 4]
}

print("BFS Traversal:")
bfs(graph, 0)
```

# DFS

```python
def dfs(graph, vertex, visited):
    visited.add(vertex)
    print(vertex, end=" ")

    for neighbour in graph[vertex]:
        if neighbour not in visited:
            dfs(graph, neighbour, visited)


# Graph representation
graph = {
    0: [1, 6],
    1: [0, 2, 3],
    2: [0, 4],
    3: [1],
    4: [1, 5],
    5: [2, 4],
    6: [] # Added node 6 to the graph
}

visited = set()

print("DFS Traversal:")
dfs(graph, 0, visited)
```
