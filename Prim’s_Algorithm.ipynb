# Prim's Algorithm

```python
# Prim's Algorithm

INF = 9999999

# Number of vertices
V = 5

# Graph represented using adjacency matrix
graph = [
    [0, 2, 0, 6, 0],
    [2, 0, 3, 8, 5],
    [0, 3, 0, 0, 7],
    [6, 8, 0, 0, 9],
    [0, 5, 7, 9, 0]
]

# Store selected vertices
selected = [False] * V

# Start from vertex 0
selected[0] = True

print("Edge : Weight")

# MST contains V-1 edges
for _ in range(V - 1):

    minimum = INF
    x = 0
    y = 0

    # Find minimum weight edge
    # connecting selected vertex to unselected vertex
    for i in range(V):
        if selected[i]:

            for j in range(V):
                if not selected[j] and graph[i][j] != 0:

                    if graph[i][j] < minimum:
                        minimum = graph[i][j]
                        x = i
                        y = j

    # Select the new vertex
    selected[y] = True

    print(f"{x} - {y} : {graph[x][y]}")
```
