```python
def coin_change(coins, amount):

    # DP table
    dp = [float('inf')] * (amount + 1)

    # 0 coins are required to make amount 0
    dp[0] = 0

    for i in range(1, amount + 1):

        for coin in coins:

            if coin <= i:
                dp[i] = min(
                    dp[i],
                    1 + dp[i - coin]
                )

    return dp


coins = [1, 2, 5]
amount = 11

dp = coin_change(coins, amount)

print("DP Table:")
print(dp)

print("Minimum number of coins =", dp[amount])
```
